```python
import pandas as pd                 # Импорт необходимых библиотек
import numpy as np
import sys

#   Получение данных из файлов
roller=pd.read_csv('C:/TMP/Подбор/roller.csv', sep=';', encoding='utf-8', header=0, index_col=False)
sleeve=pd.read_csv('C:/TMP/Подбор/sleeve.csv', sep=';', encoding='utf-8', header=0, index_col=False)
ring=pd.read_csv('C:/TMP/Подбор/ring.csv', sep=';', encoding='utf-8', header=0, index_col=False)

#   Формирование входных данных в виде датафреймов (УИН, размер, номер детали)
roller['No'] = roller['УИН'] % 1000000000
sleeve['No'] = sleeve['УИН'] % 1000000000
ring['No'] = ring['УИН'] % 1000000000

roller.rename(columns={'УИН': 'ID', 'РезИзмерения': 'Size'}, inplace=True)
sleeve.rename(columns={'УИН': 'ID', 'РезИзмерения': 'Size'}, inplace=True)
ring.rename(columns={'УИН': 'ID', 'РезИзмерения': 'Size'}, inplace=True)


#   Ввод критериев подбора
#   1. Параметры зазора: delta_min, delta_max
delta_min = float(input('Введите минимальный зазор: '))
delta_max = float(input('Введите максимальный зазор: '))

#   Создание  датафрейма сборки
#   Объединение данных из трех файлов в один датафрейм
roller_sizes = roller['Size'].to_numpy()
sleeve_sizes = sleeve['Size'].to_numpy()
ring_sizes = ring['Size'].to_numpy()

result = []

for ring_index, ring_size in enumerate(ring_sizes):
    for sleeve_index, sleeve_size in enumerate(sleeve_sizes):
        delta = ring_size - (sleeve_size + 2 * roller_sizes)
        roller_indices = np.flatnonzero(
            (delta_min <= delta) & (delta <= delta_max)
        )

        for roller_index in roller_indices:
            result.append({
                'Ring_No': ring.loc[ring_index, 'No'],
                'Sleeve_No': sleeve.loc[sleeve_index, 'No'],
                'Roller_No': roller.loc[roller_index, 'No'],
#                'Delta': delta[roller_index]
            })

likage = pd.DataFrame(result)



if len(likage) == 0:
    print('Нет вариантов подбора, удовлетворяющих критериям зазора.')
    sys.exit()
print('Количество найденных вариантов подбора:', len(likage))
#   Вывод dataframe вариантов, удовлетворяющих условию зазора (по желанию)
#likage.to_csv('C:/TMP/Подбор/likage.csv', sep=';', encoding='utf-8', index=False)

print('Зазор: ', delta_min, ' - ', delta_max, '\n')
```

```python
unic_sborka = likage[[ 'Ring_No','Sleeve_No','Roller_No' ]].nunique()
print('Количество уникальных сборок не более:', unic_sborka.min())
print('Количество уникальных колец:', unic_sborka.iloc[0])
print('Количество уникальных втулок:', unic_sborka.iloc[1])
print('Количество уникальных роликов:', unic_sborka.iloc[2])
```

```python

if unic_sborka.iloc[2] < unic_sborka.iloc[1] and unic_sborka.iloc[2] < unic_sborka.iloc[0]:
    K1 = 'Roller_No'
    if unic_sborka.iloc[1] < unic_sborka.iloc[0]:
        K2 = 'Sleeve_No'
        K3 = 'Ring_No'
    else:
        K2 = 'Ring_No'
        K3 = 'Sleeve_No'
elif unic_sborka.iloc[1] <= unic_sborka.iloc[2] and unic_sborka.iloc[1] < unic_sborka.iloc[0]:
    K1 = 'Sleeve_No'
    if unic_sborka.iloc[2] < unic_sborka.iloc[0]:
        K2 = 'Roller_No'
        K3 = 'Ring_No'
    else:
        K2 = 'Ring_No'
        K3 = 'Roller_No'
elif unic_sborka.iloc[0] <= unic_sborka.iloc[2] and unic_sborka.iloc[0] <= unic_sborka.iloc[1]:
    K1 = 'Ring_No'
    if unic_sborka.iloc[2] < unic_sborka.iloc[1]:
        K2 = 'Roller_No'
        K3 = 'Sleeve_No'
    else:
        K2 = 'Sleeve_No'
        K3 = 'Roller_No'

print('K1 =', K1, 'K2 =', K2, 'K3 =', K3)
```

```python
test_out1 = likage[K1].value_counts()       #   Подсчёт количества уникальных значений в столбце K1
ddd1 = test_out1.sort_values()              #   Сортировка значений по возрастанию, чтобы создатьсписок,
result_K1 = []                              #   куда будут записаны те K1, у которых минимальная частота
result_K1.append(ddd1.index[0])
for i in range(1, len(ddd1)):
    if ddd1.iloc[i] != ddd1.iloc[i - 1]:
        break
    result_K1.append(ddd1.index[i])

#  Датафрейм сборки с самыми редкими номерами по критерию K1 
result_s = likage[likage[K1].isin(result_K1)]
print(ddd1.index)
print(result_K1)
print(test_out1)
result_s
```

```python
test_out2 = result_s[K2].value_counts()       #   Подсчёт количества уникальных значений в столбце K2
ddd2 = test_out2.sort_values()              #   Сортировка значений по возрастанию, чтобы создатьсписок,
result_K2 = []                              #   куда будут записаны те K2, у которых минимальная частота
result_K2.append(ddd2.index[0])
for i in range(len(ddd2)):
    if ddd2.iloc[i] != ddd2.iloc[i - 1]:
        break
    result_K2.append(ddd2.index[i])

result_s2 = result_s[result_s[K2].isin(result_K2)]
print(ddd2.index)
print(result_K2)
print(test_out2)
result_s2
```

```python
test_out3 = result_s[K3].value_counts()       #   Подсчёт количества уникальных значений в столбце K3
ddd3 = test_out3.sort_values()              #   Сортировка значений по возрастанию, чтобы создатьсписок,
result_K3 = []                              #   куда будут записаны те K3, у которых минимальная частота
result_K3.append(ddd3.index[0])
for i in range(len(ddd3)):
    if ddd3.iloc[i] != ddd3.iloc[i - 1]:
        break
    result_K3.append(ddd3.index[i])

result_s3 = result_s[result_s[K3].isin(result_K3)]
print(ddd3.index)
print(result_K3)
print(test_out3)
result_s3
```

```python
#   Однозначный выбор сборки, если критерии K1, K2, K3
#   не дали однозначного результата.
#   Выбираем самую старую деталь (самый маленький номер детали)
#   из всех найденных вариантов подбора К3.

#   Однозначный выбор сборки, если критерии K1, K2, K3
#   не дали однозначного результата.
#   Выбираем самую старую деталь (самый маленький номер детали)
#   из всех найденных вариантов подбора К3.

result_out = result_s3.sort_values(by=K3, ascending=True)
out_likage = result_out.iloc[[0]]
print('Выбранная сборка: ', '\n', out_likage)
#out_likage = out_likage.drop(columns=['Delta'])
out_likage

```
